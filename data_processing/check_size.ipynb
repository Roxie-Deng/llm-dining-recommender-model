```python
import os

# 设置 Yelp 数据目录
BASE_DIR = os.path.abspath(os.path.join(os.getcwd(), ".."))
yelp_data_dir = os.path.join(BASE_DIR, "yelp_data")

# 获取所有 JSON 文件
json_files = [
    "yelp_academic_dataset_business.json",
    "yelp_academic_dataset_checkin.json",
    "yelp_academic_dataset_review.json",
    "yelp_academic_dataset_tip.json",
    "yelp_academic_dataset_user.json"
]

# 快速统计 JSON 文件大小和行数
def count_lines_fast(file_path, buffer_size=1024 * 1024):  # 1MB buffer
    """高效计算 JSON 文件的行数"""
    with open(file_path, "rb") as f:
        return sum(chunk.count(b"\n") for chunk in iter(lambda: f.read(buffer_size), b""))

file_stats = {}

for file in json_files:
    file_path = os.path.join(yelp_data_dir, file)
    
    if os.path.exists(file_path):
        file_size = os.path.getsize(file_path) / (1024 * 1024)  # MB
        num_lines = count_lines_fast(file_path)
        
        file_stats[file] = {"size_MB": round(file_size, 2), "num_records": num_lines}
    else:
        file_stats[file] = {"error": "File not found"}

# 输出统计结果
for file, stats in file_stats.items():
    print(f"{file}: {stats}")
```
